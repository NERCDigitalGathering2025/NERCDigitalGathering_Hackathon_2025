# 🎯 Bonus / Stretch Challenges

These are optional open-ended challenges for more confident participants.  
They build on the base notebooks (MIDAS, ECMWF, HadUK-Grid) and encourage cross-dataset insight, regional focus, and critical thinking.

👉 The hackathon suggests exploring both **UK-scale data** and/or focusing on **Bedfordshire** as a case region.  
([Hackathon overview](https://nercdigitalgathering2025.github.io/index.htm))  

💡 *Pick one or more of these challenges to explore deeply. There is no single right answer — creativity, visual storytelling, and clear interpretation are all encouraged!*

---

## 1. Rainfall consistency: gridded vs station

- Choose a **MIDAS rainfall station** in Bedfordshire (or elsewhere in the UK).  
- Extract the corresponding grid cell from **HadUK-Grid rainfall** for the same period.  
- Compare monthly totals, compute bias and correlation, and make a scatter plot.  
- Extend this to multiple stations to map where the largest discrepancies occur.

---

## 2. Reanalysis concordance

- Use the **ECMWF ERA5 reanalysis** precipitation.  
- Compare it with HadUK-Grid rainfall for overlapping years.  
- Questions to explore:  
  - Which months/regions show the strongest agreement or disagreement?  
  - Is there a seasonal bias in ERA5 vs HadUK?

---

## 3. Spatial scale sensitivity

- Within HadUK-Grid, pick two resolutions (e.g. 1 km vs 25 km).  
- Compute monthly or annual rainfall/temperature fields at both resolutions.  
- Quantify what detail is lost:  
  - Spatial variance  
  - Extreme percentiles (e.g. top 1% rainfall days)  
- Create side-by-side plots to show differences.

---

## 4. Regional climate time series in Bedfordshire

- Define a bounding box for **Bedfordshire**:  
  - lon: −0.89 to 0.23 °E  
  - lat: 51.95 to 52.49 °N  
  (see [hackathon overview](https://nercdigitalgathering2025.github.io/index.htm))  
- Average rainfall or temperature over this region for each time step.  
- Plot a time series, identify anomalies and trends, and compare to the UK average.

---

## 5. Extreme events and spatial footprints

- Identify the **top 5 wettest days or months** in the dataset (HadUK or ECMWF).  
- For each event, map the spatial distribution of rainfall or anomaly.  
- Compare whether extreme events are highly localized or widespread.

---

## 6. Interactive & animation

- Build an **animation** of rainfall or temperature maps across a year.  
- Or use libraries like `hvplot`, `panel`, or `plotly` for interactive plots:  
  - Allow user to pick a station, variable, or year.  
  - Switch between HadUK and ECMWF datasets.

---

## Tips & hints

- Always check **time overlap** between datasets before comparing.  
- For regional selections, use bounding boxes with `.sel()` or `.where()` in xarray.  
- Use `xarray.open_mfdataset(..., combine="by_coords")` to combine multiple NetCDF files.  
- For UK mapping, Cartopy with the **OSGB projection** is often best.
