In the hackathon, we are offering a chance to explore and interact with a range of meteorological and other data in CEDA - NERC's Centre for Environmental Data Analysis. The datasets we are looking at include ECMWF, HAD and MIDAS. These data are in different formats and structures so you can explore these differences as well.

**MIDAS** (Met Office Integrated Data Archive System): This is a database of raw weather observations from land and marine surface stations, both in the UK and globally. It contains daily, hourly and sub-hourly measurements of various parameters like temperature, rainfall, sunshine, wind, cloud cover, and present weather codes. MIDAS data is station based timeseries data in CSV format.

**ECMWF** (European Centre for Medium-Range Weather Forecasts): This organisation produces weather forecasts and climate reanalyses. ECMWF data includes estimates of atmospheric parameters like air temperature, pressure, and wind at different altitudes, as well as surface parameters like rainfall, soil moisture content, ocean-wave height, and sea-surface temperature, for the entire globe. They also have ocean reanalysis and analysis systems like OCEAN5. ECMWF has regional gridded data in NetCDF format.

**HadUK-Grid** is a dataset of gridded climate variables for the UK derived from interpolated land surface observations. It focuses on climate variables like temperature, rainfall, sunshine, mean sea level pressure, wind speed, relative humidity, vapour pressure, days of snow lying, and days of ground frost, at daily, monthly, seasonal, and annual timescales. HADUK has regional gridded data in NetCDF format.

**Soils and Land Cover** In addition to the meteorological notebooks, we are also running a fourth notebook that allows some comparison of soil types and land cover in the county of Bedfordshire. Our challenge is to undertake some spatial analysis to establish any patterns between the datasets.

## Challenge 1 — Rainfall Consistency: Gridded vs Station Data

### Background
Comparing gridded climate data with station observations is crucial for understanding data quality and spatial representativeness. This challenge explores how well gridded products capture local conditions at specific locations.

### Your Task
- Choose a **MIDAS rainfall station** in Bedfordshire (or elsewhere in the UK)
- Extract the corresponding grid cell from **HadUK-Grid rainfall** for the same period
- Compare monthly totals, compute bias and correlation, and create scatter plots
- Extend this to multiple stations to map where the largest discrepancies occur

**Success criteria:** successful data extraction, meaningful statistical comparisons, clear visualisations showing station-grid relationships, identification of spatial patterns in discrepancies.

## Challenge 2 — Reanalysis Concordance Analysis

### Background
ERA5 reanalysis provides global estimates of atmospheric and surface variables, while HadUK-Grid is derived from UK station observations. Understanding their agreement helps assess data reliability and identify systematic biases.

### Your Task
- Use the **ECMWF ERA5 reanalysis** precipitation data
- Compare it with HadUK-Grid rainfall for overlapping years
- Explore which months/regions show the strongest agreement or disagreement
- Investigate seasonal bias patterns in ERA5 vs HadUK-Grid

**Success criteria:** comprehensive temporal and spatial comparison, identification of systematic biases, clear interpretation of agreement patterns, informative visualisations.

## Challenge 3 — Spatial Scale Sensitivity Analysis

### Background
Climate data resolution affects the representation of local features and extremes. Understanding what information is lost at coarser resolutions is important for data selection and interpretation.

### Your Task
- Within HadUK-Grid, select two different resolutions (e.g. 1 km vs 25 km)
- Compute monthly or annual rainfall/temperature fields at both resolutions
- Quantify what detail is lost in terms of spatial variance and extreme percentiles
- Create side-by-side plots to show differences

**Success criteria:** systematic resolution comparison, quantification of information loss, clear visualisation of scale effects, meaningful interpretation of results.

## 💡 Tips & Hints

- Always check **time overlap** between datasets before comparing
- For regional selections, use bounding boxes with `.sel()` or `.where()` in xarray
- Use `xarray.open_mfdataset(..., combine="by_coords")` to combine multiple NetCDF files
- For UK mapping, Cartopy with the **OSGB projection** is often best
- Consider data quality flags and missing value handling
- Think about statistical significance when comparing datasets
- Use appropriate colour schemes for climate data visualisation
